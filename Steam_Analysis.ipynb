To make the numbers easier to read we replace 1000 with K and 1000000 with M

```python
#define a dictionary for the reduction
number_to_reduced_number = {
    '0 - 20000': '0 - 20K',
    '20000 - 50000': '20K - 50K',
    '50000 - 100000': '50K - 100K',
    '100000 - 200000': '100K - 200K',
    '200000 - 500000': '200K - 500K',
    '500000 - 1000000': '500K - 1M',
    '1000000 - 2000000': '1M - 2M',
    '2000000 - 5000000': '2M - 5M',
    '5000000 - 10000000': '5M - 10M',
    '10000000 - 20000000': '10M - 20M',
    '20000000 - 50000000': '20M - 50M',
    '50000000 - 100000000': '50M - 100M',
    '100000000 - 200000000': '100M - 200M'}

df['estimated_owners'] = df['estimated_owners'].map(number_to_reduced_number)
df['estimated_owners'].value_counts()
```

# Prep for Sql

Let's start by droppig some not very useful fields for analysis. For example: header_image, metacritic_url, etc.

```python
df = df.drop(columns=['about_the_game', 'screenshots', 'movies', 'header_image', 'metacritic_url', 'support_url', 'support_email', 'website', 'screenshots', 'notes'], errors='ignore')
df.shape
df.columns
```

Let's make a genres and categories table more easily query based on genre and category. 

```python
#creating genre table, and genre_to_game table
genres_df = df[['appid', 'genres']].dropna()

genres_df = genres_df.assign(
    genre=genres_df['genres'].str.split(',')
).explode('genre')

genres_df['genre'] = genres_df['genre'].str.strip()

unique_genres = genres_df['genre'].drop_duplicates().reset_index(drop=True)

genres_table = pd.DataFrame({
    'genre_id': range(1, len(unique_genres) + 1),
    'genre_name': unique_genres
})

genre_map = dict(zip(genres_table['genre_name'], genres_table['genre_id']))

genres_df['genre_id'] = genres_df['genre'].map(genre_map)

game_genres = genres_df[['appid', 'genre_id']]

df = df.drop(columns=['genres'])

#creating tag table, and tag_to_game table
tags_df = df[['appid', 'tags']].dropna()

tags_df = tags_df.assign(
    tags=tags_df['tags'].str.split(',')
).explode('tags')

tags_df['tags'] = tags_df['tags'].str.strip()

unique_tags = tags_df['tags'].drop_duplicates().reset_index(drop=True)

tag_table = pd.DataFrame({
    'tag_id': range(1, len(unique_tags) + 1),
    'tag_name': unique_tags
})

tag_map = dict(zip(tag_table['tag_name'], tag_table['tag_id']))

tags_df['tag_id'] = tags_df['tags'].map(tag_map)

game_tags = tags_df[['appid', 'tag_id']]

df = df.drop(columns=['tags'])
```

make a set of tables for `supported_languages`

```python
# languages
languages_df = df[['appid', 'supported_languages']].dropna()

languages_df = languages_df.assign(
    supported_languages=languages_df['supported_languages'].str.strip("[]").str.split("', '")
).explode('supported_languages')

languages_df['supported_languages'] = (
    languages_df['supported_languages']
    .str.strip("'")
    .str.strip()
)

languages_df['supported_languages'] = languages_df['supported_languages'].str.strip()

unique_languages = languages_df['supported_languages'].drop_duplicates().reset_index(drop=True)

languages_table = pd.DataFrame({
    'languages_id': range(1, len(unique_languages) + 1),
    'languages_name': unique_languages
})

languages_map = dict(zip(languages_table['languages_name'], languages_table['languages_id']))

languages_df['languages_id'] = languages_df['supported_languages'].map(languages_map)

game_languages = languages_df[['appid', 'languages_id']]
```

It seems supported languages has some messy data. We will look though all languages and keep make a .csv with plausible languages. We will redo the conversion with the new list of languages.

```python
language_names = [
    'English', 'French', 'German', 'Russian', 'Korean', 'Italian',
    'Spanish - Spain', 'Polish', 'Portuguese - Portugal', 'Turkish',
    'Bulgarian', 'Simplified Chinese', 'Traditional Chinese', 'Danish',
    'Spanish - Latin America', 'Finnish', 'Greek', 'Dutch', 'Hungarian',
    'Japanese', 'Norwegian', 'Portuguese - Brazil', 'Romanian', 'Swedish',
    'Thai', 'Czech', 'Ukrainian', 'Vietnamese', 'Arabic', 'Belarusian',
    'Malay', 'Indonesian', 'Afrikaans', 'Albanian', 'Amharic', 'Armenian',
    'Assamese', 'Azerbaijani', 'Bangla', 'Basque', 'Bosnian', 'Catalan',
    'Cherokee', 'Croatian', 'Dari', 'Estonian', 'Filipino', 'Galician',
    'Georgian', 'Gujarati', 'Hausa', 'Hebrew', 'Hindi', 'Icelandic',
    'Igbo', 'Irish', 'Kannada', 'Kazakh', 'Khmer', 'Kinyarwanda',
    'Konkani', 'Kyrgyz', 'Latvian', 'Lithuanian', 'Luxembourgish',
    'Macedonian', 'Malayalam', 'Maltese', 'Maori', 'Marathi', 'Mongolian',
    'Nepali', 'Odia', 'Persian', 'Punjabi (Gurmukhi)',
    'Punjabi (Shahmukhi)', 'Quechua', 'Scots', 'Serbian', 'Sindhi',
    'Sinhala', 'Slovak', 'Slovenian', 'Sorani', 'Sotho', 'Swahili',
    'Tajik', 'Tamil', 'Tatar', 'Telugu', 'Tigrinya', 'Tswana', 'Turkmen',
    'Urdu', 'Uyghur', 'Uzbek', 'Valencian', 'Welsh', 'Wolof', 'Xhosa',
    'Yoruba', 'Zulu'
]

languages_df = languages_df.assign(
    languages=languages_df['supported_languages'].apply(
        lambda x: [language for language in language_names if language in str(x)]
    )
).explode('languages')

unique_languages = languages_df['languages'].drop_duplicates().reset_index(drop=True)

languages_table = pd.DataFrame({
    'languages_id': range(1, len(unique_languages) + 1),
    'languages_name': unique_languages
})

languages_map = dict(zip(languages_table['languages_name'], languages_table['languages_id']))

languages_df['languages_id'] = languages_df['supported_languages'].map(languages_map)

game_languages = languages_df[['appid', 'languages_id']].drop_duplicates()

game_languages = game_languages.dropna()

game_languages['languages_id'] = (
    game_languages['languages_id'].astype(int)
)

df = df.drop(columns=['supported_languages'])
```
